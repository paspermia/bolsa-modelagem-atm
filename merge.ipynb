```python
import xarray as xr
import pandas as pd
```

DIA 3

```python
#dia 3

# Abrir o arquivo do MERGE
ds_merge = xr.open_dataset('MERGE_CPTEC_20230903.grib2', engine='cfgrib')

# Verificar e ajustar Coordenadas
ds_merge.coords['longitude'] = (ds_merge.coords['longitude'] + 180) % 360 - 180
ds_merge = ds_merge.sortby(ds_merge.longitude)

print("Arquivo aberto e coordenadas corrigidas com sucesso!")
print(ds_merge)
```

```python
# Extração da chuva do MERGE usando Interpolação Bilinear

# definição das coordenadas das estações
estacoes = {
    'Bento Gonçalves': {'lat': -29.16, 'lon': -51.52},
    'Caxias do Sul':   {'lat': -29.17, 'lon': -51.18},
    'Santa Maria':     {'lat': -29.68, 'lon': -53.81}
}

print("Extraindo dados de chuva do MERGE por interpolação linear...\n")

for nome_cidade, coord in estacoes.items():
    chuva_interp = ds_merge['rdp'].interp(
        latitude=coord['lat'],
        longitude=coord['lon'],
        method='linear'
    ).values

    print(f"Chuva pelo MERGE em {nome_cidade}: {float(chuva_interp):.2f} mm")
```

```python
# Dados estruturados (Janela 12h-12h UTC)
dados = {
    'Data': ['2023-09-03', '2023-09-03', '2023-09-03'],
    'Cidade': ['Bento Gonçalves', 'Caxias do Sul', 'Santa Maria'],
    'Precip_INMET_mm': [2.8, 1.8, 4.4],
    'Precip_MERGE_mm': [32.94, 33.90, 12.24]
}

df = pd.DataFrame(dados)

# Calcular a diferença (MERGE - INMET)
df['Diferenca_mm'] = df['Precip_MERGE_mm'] - df['Precip_INMET_mm']

# Configurar espaçamento automático das colunas
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)

# Mostrar o resultado
print("Resultado da Comparação:\n")
print(df.to_string(index=False, col_space=18))
```
